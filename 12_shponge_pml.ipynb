## Борьба с отражениями на границах модели: Shponge Layer vs PML

В этом ноутбуке с помощью 1D FD-моделирования рассмотрим реализацию и отличия двух популярных граничных условий: shponge layer (он же absorbing boundary) и PML (perfectly matched layer) в простейшей реализации unsplit PML.

Идея shponge layer: домножить поле на профиль затухания σ на краях модели:
$$\frac{\partial^2 p}{\partial t^2} = c^2 \nabla^2 p - \sigma \frac{\partial p}{\partial t}$$

**Механизм работы:** Коэффициент затухания σ увеличивается от нуля в рабочей области до максимума на границах по квадратичному профилю. Член $-σ \frac{\partial p}{\partial t}$ создает экспоненциальное затухание волн в PML-зоне.

**Проблема такого подхода:** Из-за изменения поля на краях получаются разрывы градиента, что эквивалентно разрыву импеданса, который рождает отражения. Даже при плавном профиле σ отражения остаются, хотя и уменьшаются с увеличением толщины слоя.

Идея PML: скомпенсировать скачки градиента с помощью ввода вспомогательного поля ψ, которое "запоминает" градиент давления:
$$\frac{\partial^2 p}{\partial t^2} = c^2 \frac{\partial}{\partial x}\left(\frac{\partial p}{\partial x} - \psi\right) - \sigma \frac{\partial p}{\partial t}$$

**Два механизма поглощения:** В PML работают два согласованных механизма:
1. **Модификация градиента** (через ψ): $c^2 \frac{\partial}{\partial x}\left(\frac{\partial p}{\partial x} - \psi\right)$ — компенсирует разрывы градиента
2. **Затухание по времени** (через σ): $- \sigma \frac{\partial p}{\partial t}$ — поглощает энергию (как в Sponge Layer)

**Связь с классическим описанием через растяжение координат:** Классическое описание PML основано на комплексном растяжении координат. В частотной области координата $x$ заменяется на растянутую координату $\tilde{x} = x + \frac{i}{\omega}\int_0^x \sigma(\xi) d\xi$, где $i$ — мнимая единица, $\omega$ — частота. Это эквивалентно замене производной:
$$\frac{\partial}{\partial x} \rightarrow \frac{1}{1 + i\sigma/\omega} \frac{\partial}{\partial x}$$

В временной области это растяжение координат реализуется через введение вспомогательного поля ψ, которое "запоминает" градиент. Таким образом, наше описание через ψ эквивалентно классическому описанию через комплексное растяжение координат, но сформулировано во временной области без использования преобразования Фурье.

**Компенсация:** В отличие от Sponge Layer, где затухание по времени создает отражения, в PML модификация градиента компенсирует эти отражения. Можно переписать уравнение как:
$$\frac{\partial^2 p}{\partial t^2} = c^2 \frac{\partial^2 p}{\partial x^2} - c^2 \frac{\partial \psi}{\partial x} - \sigma \frac{\partial p}{\partial t}$$
где член $-c^2 \frac{\partial \psi}{\partial x}$ компенсирует отражения, вызванные членом $- \sigma \frac{\partial p}{\partial t}$.

**Механизм компенсации:** В PML-зоне поле ψ обновляется как $\frac{\partial \psi}{\partial t} = \sigma_{flux} \left(\frac{\partial p}{\partial x} - \psi\right)$, где $\sigma_{flux}$ связан с $\sigma$ через условие "perfect matching". В результате $\psi \approx \frac{\partial p}{\partial x}$, и эффективный градиент $\frac{\partial p}{\partial x} - \psi \approx 0$, что устраняет разрывы градиента на границе рабочей области и PML-зоны.

**Связь с затуханием:** Параметры $\sigma$ (затухание по времени) и $\sigma_{flux}$ (скорость накопления в ψ) должны быть согласованы для идеального поглощения без отражений. Это условие "perfect matching" — ключевое отличие PML от Sponge Layer.

**Результат:** PML эффективно поглощает волны без отражений за счет согласованной модификации пространственной производной (через ψ) и временного затухания (через σ), где первый механизм компенсирует отражения, создаваемые вторым.


```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
```

### 1. Создание модели и исчтоника для 1D симуляции

```python
# grid
nx = 200
dx = 10.0
dt = 0.0005
nt = 2000

c = 2000.0
x = np.arange(nx) * dx

# wavefield
p = np.zeros(nx)
p_old = np.zeros(nx)
p_new = np.zeros(nx)

# source
src = nx // 2
f0 = 20.0

# snapshot saving
snap_step = 25  # сохранять каждый N-й снапшот
```

```python
def ricker(t, f0):
    a = np.pi * f0 * (t - 1.0 / f0)
    return (1 - 2*a**2) * np.exp(-a**2)
```

### 2. Функция для обновления давления с учетом затухания по времени

```python
def update_pressure_with_damping(p, p_old, laplacian, sigma, c, dt):
    """
    Обновляет давление с учетом затухания по времени (semi-implicit схема).
    
    Реализует уравнение: (p_new - 2p + p_old)/dt² = c²·laplacian - σ·(p_new - p_old)/(2·dt)
    
    Параметры:
    ----------
    p : np.ndarray
        Текущее давление
    p_old : np.ndarray
        Предыдущее давление
    laplacian : np.ndarray
        Лапласиан давления (может быть обычным или модифицированным через PML)
    sigma : np.ndarray
        Профиль коэффициента затухания
    c : float
        Скорость волны
    dt : float
        Шаг по времени
    
    Возвращает:
    -----------
    p_new : np.ndarray
        Новое давление после обновления с затуханием
    """
    # denom (знаменатель) = коэффициент при p_new после преобразования уравнения
    # В рабочей области (σ=0): denom = 1 (обычное волновое уравнение)
    # В зонах поглощения (σ>0): denom > 1 (усиливает затухание)
    denom = 1.0 + sigma * dt / 2.0
    
    # numer (числитель) = правая часть уравнения после преобразования
    # Содержит: текущее состояние (2p - p_old), лапласиан (c²·dt²·laplacian),
    # и поправку на затухание (σ·dt/2·p_old)
    numer = (2.0 * p - p_old +
             (c * dt)**2 * laplacian +
             sigma * dt / 2.0 * p_old)
    
    # Деление numer/denom решает уравнение относительно p_new
    # В рабочей области: p_new ≈ numer (denom ≈ 1)
    # В зонах поглощения: p_new < numer (denom > 1), что создает затухание
    return numer / denom
```

### 3. Функция для Shponge Layer

```python
def sponge_modelling(npml, R_sponge):
    """
    Выполняет FD моделирование с использованием Sponge Layer поглощения границ.
    
    Параметры:
    ----------
    npml : int
        Размер зоны PML с каждой стороны
    R_sponge : float
        Теоретический коэффициент отражения для Sponge layer
    
    
    Возвращает:
    -----------
    snapshots : list
        Список снапшотов волнового поля для анимации
    sigma : numpy.ndarray
        Профиль коэффициента затухания sigma
    """
    # Параметры сетки, среды и источника берутся из глобальных переменных
    # (nx, dx, dt, nt, c, f0, snap_step, src)
    
    # Инициализация волнового поля
    p = np.zeros(nx)
    p_old = np.zeros(nx)
    p_new = np.zeros(nx)
    
    # Создание профиля sigma
    sigma = np.zeros(nx)
    sigma_max = -(np.log(R_sponge) * 3 * c) / (2 * npml * dx)
    
    # Проверка на устойчивость
    if sigma_max * dt > 1.0:
        sigma_max = 1.0 / dt
        print("Warning: Sigma ограничена условием устойчивости")
    
    # Квадратичный профиль sigma
    for i in range(npml):
        val = sigma_max * ((npml - i) / npml)**2
        sigma[i] = val
        sigma[-i-1] = val
    
    # Список для сохранения снапшотов
    snapshots = []
    
    # FD моделирование
    for it in range(nt):
        t = it * dt
        
        # Laplacian (обычный 3-точечный)
        lap = np.zeros(nx)
        lap[1:-1] = (p[2:] - 2*p[1:-1] + p[:-2]) / dx**2
        
        # Обновление давления с затуханием (semi-implicit схема)
        # Используется общая функция update_pressure_with_damping
        p_new = update_pressure_with_damping(p, p_old, lap, sigma, c, dt)
        
        # Источник
        p_new[src] += ricker(t, f0) * (dt**2)
        
        p_old[:] = p
        p[:] = p_new
        
        # Сохранение снапшота
        if it % snap_step == 0:
            snapshots.append(p.copy())
    
    return snapshots, sigma
```

### 4. Функция для Unsplit PML


```python
def pml_modelling(npml, R_theo):
    """
    Выполняет FD моделирование с использованием 1D PML (unsplit).

    Параметры
    ---------
    npml : int
        Размер зоны PML с каждой стороны.
    R_theo : float
        Теоретический коэффициент отражения для PML.

    Возвращает
    -----------
    snapshots : list of np.ndarray
        Список снапшотов волнового поля p для анимации.
    sigma : np.ndarray
        Профиль sigma в узлах (для затухания p).
    sigma_flux : np.ndarray
        Профиль sigma между узлами (интерполяция для psi).
    """
    # Параметры сетки и среды берутся из глобальных переменных
    # (nx, dx, dt, nt, c, f0, snap_step, src)

    # Расчет sigma под параметры сетки для лучшего поглощения
    sigma = np.zeros(nx)
    sigma_max = -(np.log(R_theo) * 3 * c) / (2 * npml * dx)

    for i in range(npml):
        d = (npml - i) / npml
        val = sigma_max * d**2
        sigma[i] = val
        sigma[-i-1] = val

    # Инициализация полей
    p = np.zeros(nx)
    p_old = np.zeros(nx)
    p_new = np.zeros(nx)

    # psi "запоминает" градиент давления в PML-зоне (между узлами)
    psi = np.zeros(nx - 1)
    # sigma_flux для psi также между узлами (интерполяция sigma)
    sigma_flux = 0.5 * (sigma[:-1] + sigma[1:])

    snapshots = []
    psi_snapshots = []
    pressure_gradient_snapshots = []
    effective_gradient_snapshots = []
    laplacian_snapshots = []

    for it in range(nt):
        t = it * dt

        # ШАГ 1: Градиент давления (∂p/∂x)
        pressure_gradient = (p[1:] - p[:-1]) / dx

        # ШАГ 2: psi "запоминает" градиент
        psi = psi + dt * sigma_flux * (pressure_gradient - psi)

        # ШАГ 3: Эффективный градиент (скомпенсированный)
      
        effective_gradient = pressure_gradient - psi

        # ШАГ 4: Дивергенция эффективного градиента = ∂²p/∂x²
        laplacian = np.zeros(nx)
        laplacian[1:-1] = (effective_gradient[1:] - effective_gradient[:-1]) / dx

        # ШАГ 5: Волновое уравнение ∂²p/∂t² = c² · ∂²p/∂x² - σ · ∂p/∂t
        # (semi-implicit схема с затуханием sigma)
        # Используется общая функция update_pressure_with_damping
        p_new = update_pressure_with_damping(p, p_old, laplacian, sigma, c, dt)

        # Источник
        p_new[src] += ricker(t, f0) * (dt**2)

        p_old[:] = p
        p[:] = p_new

        if it % snap_step == 0:
            snapshots.append(p.copy())
            psi_snapshots.append(psi.copy())
            pressure_gradient_snapshots.append(pressure_gradient.copy())
            effective_gradient_snapshots.append(effective_gradient.copy())
            laplacian_snapshots.append(laplacian.copy())

    return snapshots, sigma, psi_snapshots, pressure_gradient_snapshots, effective_gradient_snapshots, laplacian_snapshots
```

### 5. Функция для создания сравнительной анимации волновых полей

```python
def create_animation(snap_sponge, snap_pml, suptitle=''):
    """
    Создает анимацию для сравнения Sponge Layer и PML моделирования.
    
    Параметры:
    ----------
    snap_sponge : list
        Список снапшотов для Sponge Layer моделирования
    snap_pml : list
        Список снапшотов для PML моделирования
    suptitle : str, optional
        Общий заголовок для всех subplots
    
    Возвращает:
    -----------
    HTML объект с анимацией
    """
    # Параметры сетки берутся из глобальных переменных
    # (x, dx, dt, snap_step, n_pml, nx)

    global n_pml
    global x
    global dx
    global dt
    global snap_step
    
    # Создаем фигуру с двумя подграфиками (один под другим)
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
    
    # Устанавливаем общий заголовок
    if suptitle:
        fig.suptitle(suptitle, fontsize=14)
    
    # Инициализация графиков
    line1, = ax1.plot([], [], 'b-', linewidth=2, label='Sponge')
    line2, = ax2.plot([], [], 'r-', linewidth=2, label='PML')
    
    ymin_sponge, ymax_sponge = -np.max(snap_sponge), np.max(snap_sponge)
    ymin_pml, ymax_pml = -np.max(snap_pml), np.max(snap_pml)
    
    # Настройка осей
    ax1.set_xlim(0, x[-1])
    ax1.set_ylim(ymin_sponge, ymax_sponge)
    ax1.set_xlabel('x, m')
    ax1.set_ylabel('Pressure')
    ax1.set_title('SPONGE LAYER')
    ax1.grid(True)
    ax1.legend()
    
    ax2.set_xlim(0, x[-1])
    ax2.set_ylim(ymin_pml, ymax_pml)
    ax2.set_xlabel('x, m')
    ax2.set_ylabel('Pressure')
    ax2.set_title('PML')
    ax2.grid(True)
    ax2.legend()
    
    # Определяем границы зон PML
    pml_left = n_pml * dx  # правая граница левой зоны PML
    pml_right = (nx - n_pml) * dx  # левая граница правой зоны PML
    
    # Выделяем зоны PML цветом на обоих графиках
    ax1.axvspan(0, pml_left, alpha=0.2, color='yellow', label='PML zone')
    ax1.axvspan(pml_right, x[-1], alpha=0.2, color='yellow')
    ax2.axvspan(0, pml_left, alpha=0.2, color='yellow', label='PML zone')
    ax2.axvspan(pml_right, x[-1], alpha=0.2, color='yellow')
    
    # Текстовый элемент для отображения номера кадра
    text1 = ax1.text(0.02, 0.95, '', transform=ax1.transAxes, fontsize=12,
                     verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
    text2 = ax2.text(0.02, 0.95, '', transform=ax2.transAxes, fontsize=12,
                     verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
    
    def animate(frame):
        # Обновляем данные для обоих графиков
        line1.set_data(x, snap_sponge[frame])
        line2.set_data(x, snap_pml[frame])
        
        # Получаем номер итерации
        it = frame * snap_step
        
        # Обновляем текст с номером итерации
        text1.set_text(f'Time: {it*dt:.3f} s')
        text2.set_text(f'Time: {it*dt:.3f} s')
        
        return line1, line2, text1, text2
    
    # Создаем анимацию
    anim = FuncAnimation(fig, animate, frames=len(snap_sponge), interval=100, blit=True, repeat=True)
    
    plt.tight_layout()
    plt.close()
    return HTML(anim.to_jshtml())
```

### 6. Эксперименты с разным n_pml

```python
n_pml = 5
R_theo = 1e-10 # тут просто какое-то очень маленькое число

snap_sponge, sigma = sponge_modelling(n_pml, R_theo)
snap_pml, sigma_pml, psi_snapshots, pressure_gradient_snapshots, \
                 effective_gradient_snapshots, laplacian_snapshots = pml_modelling(n_pml, R_theo)

create_animation(snap_sponge, snap_pml, suptitle=f'Sponge Layer vs PML: n_pml = {n_pml}, R_theo = {R_theo}')
```
